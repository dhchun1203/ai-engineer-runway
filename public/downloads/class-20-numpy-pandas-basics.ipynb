### 배열로 바꾸면 곱하기 한 번이 모든 칸에


## 3. 조건으로 고르기와 배열의 한계

### True/False 줄을 대괄호에 넣기

**나와야 하는 것:** `False`, `True` 12개짜리 줄과 `array([15000, 32000, 10900, 18500])`

````python
print(arr >= 10000)
arr[arr >= 10000]
````

### 배열은 한 가지 종류만 담는다

**나와야 하는 것:** 숫자 3200, 1500이 따옴표 붙은 글자 `'3200'`, `'1500'`이 되고, `값의 종류: <U21`

````python
mixed = np.array([["편의점", 3200],
                  ["버스",   1500]])
print(mixed)
print("값의 종류:", mixed.dtype)
````

### 가게와 분류는 따로 리스트로, 식비 합계는 반복문으로

**나와야 하는 것:** `24200`. 다음 절에서 이 리스트들로 표를 만들어요.

````python
shops = ["편의점", "버스", "영화관", "분식집", "온라인쇼핑", "카페",
         "음원앱", "버스", "편의점", "웹툰", "온라인쇼핑", "분식집"]

categories = ["식비", "교통", "문화", "식비", "쇼핑", "식비",
              "문화", "교통", "식비", "문화", "쇼핑", "식비"]

food = 0
for c, a in zip(categories, amounts):
    if c == '식비':
        food += a

print(food)
````

## 4. DataFrame 만들기

**나와야 하는 것:** 0~11번 행, `day`, `shop`, `category`, `amount` 네 열의 표. 첫 행은 `2  편의점  식비  3200`이에요.

````python
spend = pd.DataFrame({
    "day": [2, 3, 4, 5, 7, 9, 12, 14, 16, 19, 20, 25],
    "shop": shops,
    "category": categories,
    "amount": amounts
})

spend
````

### shape, head, info

**나와야 하는 것:** `(12, 4)`와 앞 3행. 다음 칸의 `info()`는 열마다 `12 non-null`이고 `day`, `amount`는 `int64`, `shop`, `category`는 `str`이에요.

````python
print(spend.shape)
spend.head(3)
````

````python
spend.info()
````

## 5. 열 꺼내기: 한 겹은 Series, 두 겹은 표

**나와야 하는 것:** 0~11번 금액이 세로로 찍히고 `Name: amount, dtype: int64`, 그 아래 `합계: 106600`, `평균: 8883.33`

````python
print(spend['amount'])

print('합계:', spend['amount'].sum())
print('평균:', round(spend['amount'].mean(), 2))
````

**나와야 하는 것:** `shop`, `amount` 두 열만 있는 표

````python
spend[['shop', 'amount']]
````

**나와야 하는 것:** `(12,)`와 `(12, 1)`

````python
print(spend["amount"].shape)     # 대괄호 한 겹: 한 줄
print(spend[["amount"]].shape)   # 대괄호 두 겹: 표
````

## 6. 행 거르기

### 조건 하나

**나와야 하는 것:** 첫 칸은 `True`가 2, 6, 9번에만 있는 줄. 둘째 칸은 영화관, 음원앱, 웹툰 3행. 셋째 칸은 영화관, 온라인쇼핑, 음원앱, 온라인쇼핑 4행.

````python
spend["category"] == "문화"
````

````python
spend[spend["category"] == "문화"]
````

````python
spend[spend['amount'] >= 10000]
````

### 조건 둘: &로 잇고 조건마다 괄호

**나와야 하는 것:** 3번(분식집 6500)과 11번(분식집 7000) 두 행

````python
spend[(spend["category"] == '식비') & (spend["amount"] >= 5000)]
````

**해 보기:** 수업에서 만난 오류 두 개예요. 한 줄씩 `#`을 지우고 실행해 보세요.

- `and`를 쓰면 `ValueError: The truth value of a Series is ambiguous. ...`

## 7. 정렬, 새 열, 한 칸 고치기

**나와야 하는 것:** 4번(32000), 10번(18500), 2번(15000) 세 행. 행 번호도 함께 따라와요.

````python
spend.sort_values('amount', ascending=False).head(3)
````

**나와야 하는 것:** `1066`과, 마지막에 `point` 열(32, 15, 150)이 붙은 앞 3행

````python
spend['point'] = spend['amount'] // 100   # 100원마다 1포인트
print(spend['point'].sum())
spend.head(3)
````

**해 보기:** 아래 칸의 `#`을 지우고 실행해 보세요. 꺼낸 열에 대괄호를 한 번 더 붙여 고치면 `ChainedAssignmentError` 경고만 나오고 2번 행의 금액은 15000 그대로예요. 확인했으면 다시 `#`을 붙여요.

````python
# spend["amount"][2] = 14000
# spend.head(3)
````

**나와야 하는 것:** 2번 행이 `14000`, `140`으로 바뀐 앞 3행. `point`는 계산해 둔 열이라 금액을 고친 뒤 다시 계산해야 바뀌어요.

````python
spend.loc[2, 'amount'] = 14000
spend['point'] = spend['amount'] // 100
spend.head(3)
````

## 8. groupby로 분류별 합계

**나와야 하는 것:** 첫 칸은 교통 3000, 문화 27900, 쇼핑 50500, 식비 24200 (이름 순). 둘째 칸은 큰 순으로 쇼핑 50500, 문화 27900, 식비 24200, 교통 3000. 셋째 칸은 식비 5, 문화 3, 교통 2, 쇼핑 2.

문화 27900은 7절에서 영화관을 14000으로 고친 뒤의 합계예요(14000 + 10900 + 3000).

````python
spend.groupby('category')['amount'].sum()
````

````python
spend.groupby('category')['amount'].sum().sort_values(ascending=False)
````

````python
spend['category'].value_counts()
````
