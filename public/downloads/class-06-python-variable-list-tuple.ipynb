## 3. 문자열 포맷

세 방식을 차례로 돌려 봐요. 실무에서는 f-string을 주로 써요.

````python
print("선발 출전: %s, %d" % ("손흥민", 20))
print("선발 출전: [%5d]" % (20))
print("소수점 표기 : %.2f" % (3.141592))
````

````python
name = "Cake"
price = 6500
qty = 2

format_str = "주문내역: {0} {1} -> 총 {2}원".format(name, qty, price)
print(format_str)

format_str = f"주문내역: {name} {price} -> 총 {(price * qty):>10,}원"
print(format_str)

print(f"금액: {15000:>8,}원")
````

````python
orders = [
    ("Latte", 5000, 2),
    ("Americano", 3000, 1),
    ("Cake", 6000, 1)
]
print("=" * 34)
print(f"{'메뉴명':<10} {'단가':>8} {'금액':>9}")
print("-" * 34)

total_amount = 0

for menu, price, qty in orders:
    subtotal = price * qty
    total_amount += subtotal
    print(f"{menu:<10} {price:>12,} {subtotal:>10,}")

print("-" * 34)
print(f"{'합계금액':<10}{total_amount:>20,}원")
````

## 4. 리스트와 튜플

````python
shopping_list = ["apple", "banana", "cherry"]
print(shopping_list)
print(shopping_list[0])

shopping_list[0] = "kiwi"
print(shopping_list[0])
print(shopping_list)
````

````python
fruits = ["apple"]
fruits.append("banana")
print(fruits)

fruits.insert(0, "kiwi")
print(fruits)

fruits.remove("apple")
print(fruits)

# fruits.remove("grape")  # 주석을 풀어 보세요: ValueError
````

````python
colors = ["red", "green", "blue"]
print(colors[0])
print(colors[-1])
````

슬라이싱은 칸의 마지막 줄만 자동으로 나와요. 줄을 지우거나 순서를 바꿔 가며 하나씩 봐요.
