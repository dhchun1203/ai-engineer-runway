## 2. 선형회귀 복습: 공부 시간 5명

## 4. 손실 함수: 틀린 정도를 MSE 하나로

손실 함수는 예측이 정답과 얼마나 다른지를 숫자 하나로 나타내요. 선형회귀는 보통 MSE를 써요.

MSE = (1/n) × Σ (실제 yᵢ − 예측 ŷᵢ)²

경사하강법은 이 숫자가 줄어드는 쪽으로 w, b를 조금씩 고쳐요. 아래 칸은 수업에 없던 칸으로, w와 b에 따라 손실이 어떻게 달라지는지 숫자로 봐요.

**나와야 하는 것:**
```
4428.2
12.2
3.2
```

````python
import numpy as np

hours = np.array([1, 2, 3, 4, 5])
scores = np.array([47, 53, 65, 73, 87])

# 손실(MSE) 계산 함수
def loss(w, b):
    return ((scores - (w * hours + b)) ** 2).mean()

print(loss(0, 0))    # 시작점
print(loss(12, 30))  # 가까워졌다
print(loss(10, 35))  # 가장 낮은 곳
````

## 5. SGDRegressor로 한 걸음씩 배우기

### 학습 데이터 준비

**나와야 하는 것:** 2절과 같은 5행 표

````python
from sklearn.linear_model import LinearRegression
# 배치 경사하강법 (BGD) : 전체 데이터를 사용, 시간, 리소스 많이 사용
# 확률적 경사하강법 (SGD) : 확률적을 1개의 데이터를 추출해서 하강 방법
from sklearn.linear_model import SGDRegressor
from sklearn.metrics import mean_squared_error

# 학습 데이터 생성
df = pd.DataFrame({
    "hours": [1, 2, 3, 4, 5],
    "scores": [47, 53, 65, 73, 87]
})

# 독립변수 (입력변수 : Feature)
X = df[["hours"]]
# 종속변수 (정답 : Target)
y = df["scores"]

df
````

### 목표 먼저 재 두기

**나와야 하는 것:**
```
Weight: 10.00
Bias: 35.00
MSE: 3.2000
```

````python
# 선형회귀
model = LinearRegression()
# 모델 학습
model.fit(X, y)
# 예측값
y_pred = model.predict(X)
# 결과값 : 경사하강법으로 도달하려는 목표
print(f"Weight: {model.coef_[0]:.2f}")
print(f"Bias: {model.intercept_:.2f}")
print(f"MSE: {mean_squared_error(y, y_pred):.4f}")
````

### 확률적 경사하강법 모델 생성

**나와야 하는 것:** 아무것도 안 나와요. 학습 칸을 다시 돌리고 싶으면 이 칸부터 다시 실행해요(안 그러면 이어서 배워 숫자가 달라져요).

````python
sgd = SGDRegressor(
    loss="squared_error",       # 제곱 오차 기반 손실함수
    learning_rate="constant",   # 학습률을 일정하게 유지
    eta0=0.01,                  # 초기 학습률
    max_iter=1000,              # 학습횟수 fit() 반복 회수
    shuffle=False,              # 학습 데이터 섞지 않음
    random_state=42             # 난수 시드 설정
)
````

### 1000 epoch 학습하고 기록하기

**나와야 하는 것:** `[1000 rows x 4 columns]` 표. 첫 줄 `1   8.455225   2.731163  1369.816039`, 마지막 줄 `1000  10.129335  34.741238     3.250159`

````python
# 전체 데이터를 기준으로 총 몇회 반복
epochs = 1000

# 학습과정 저장 리스트
history = []

for epoch in range(epochs):
    # 전체 데이터 1회 학습 (1 epoch)
    sgd.partial_fit(X, y)

    # 1회 학습 후 예측
    y_pred = sgd.predict(X)

    # 현재 모델의 MSE 계산
    mse = mean_squared_error(y, y_pred)

    # 현재 학습 결과 저장(누적)
    history.append({
        "epoch": epoch + 1,
        "weight": sgd.coef_[0],
        "bias": sgd.intercept_[0],
        "mse": mse
    })

# 학습 기록 => DataFrame 으로 변환
history_df = pd.DataFrame(history)
history_df
````

### 고른 epoch만 보기

**나와야 하는 것:** 8행 표. mse가 1369.816039 → 167.673523 → 141.377004 → 72.111791 → 32.278720 → 4.254356 → 3.302220 → 3.250159로 줄어요.

````python
steps = [1, 5, 10, 50, 100, 300, 500, 1000]
history_df[history_df["epoch"].isin(steps)]
````

### 1 epoch를 손으로 따라가 보기 (수업에 없던 칸)

w, b를 0에서 시작해 학생 다섯 명을 차례로 보며 고쳐요. `alpha`는 SGDRegressor의 기본 규제 세기예요.

**나와야 하는 것:** `8.455224781754193 2.731163490747725` (위 표의 1 epoch와 같아요)

````python
w = 0.0
b = 0.0
eta = 0.01     # 학습률 eta0
alpha = 0.0001 # SGDRegressor 기본 규제 세기

for xi, yi in zip([1, 2, 3, 4, 5], [47, 53, 65, 73, 87]):
    error = (w * xi + b) - yi
    w = w * (1 - eta * alpha) - eta * error * xi
    b = b - eta * error

print(w, b)
````

## 6. 학습 곡선 그리기

### 손실 함수 차트

수업 코드에서 가로축 이름을 `"Loss"`에서 `"Epoch"`으로 고치고, 범례가 나오도록 `plt.legend()`를 더했어요.

**나와야 하는 것:** 왼쪽 위(1400 근처)에서 뚝 떨어져 바닥에 붙는 파란 곡선, 범례 `학습 손실(Training Loss)`

````python
plt.figure(figsize=(10,5))

# Epoch에 따른 MSE 변화
plt.plot(
    history_df["epoch"],
    history_df["mse"],
    color="blue",
    label="학습 손실(Training Loss)"
)
plt.xlabel("Epoch")
plt.ylabel("MSE")
plt.grid(alpha=0.3)
plt.legend()
plt.show()
````

### 기울기와 절편 변화 시각화

**나와야 하는 것:** 그래프 두 칸. 왼쪽(Weight)은 처음에 17 근처까지 치솟았다가 10으로 천천히 내려오고, 오른쪽(Bias)은 0에서 35를 향해 올라가요.

````python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# 기울기 변화
axes[0].plot(
    history_df["epoch"],
    history_df["weight"],
    color="blue"
)
axes[0].grid(alpha=0.3)
axes[0].set_ylabel("Weight")

# 절편 변화
axes[1].plot(
    history_df["epoch"],
    history_df["bias"],
    color="blue"
)
axes[1].grid(alpha=0.3)
axes[1].set_ylabel("Bias")

plt.show()
````
