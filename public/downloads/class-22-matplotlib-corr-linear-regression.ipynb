## 4. seed로 가상 오피스텔 200건 만들기

### 수업 셀 5

**나와야 하는 것:** seed가 같아서 누구나 똑같은 표가 나와요.

```text
        area    price
count  200.0    200.0
mean    52.8  29138.4
std     18.6   9112.1
min     22.0  11400.0
25%     37.4  21615.0
50%     52.0  28965.0
75%     67.4  36377.5
max     84.9  58460.0
```

````python
# [셀 5] 가상 오피스텔 데이터 만들기
# 시드(seed) : 난수 발생시 참고하는 데이터, 동일한 seed 값을 적용하면
# 항상 동일한 난수가 발생
rng = np.random.default_rng(40)

n = 200
area = rng.uniform(20, 85, n) # 20 이상 85 미만에서 균등하게 n개 만큼 난수 발생

# 선형회귀
# y = wx + b
# w: 기울기 450
# b: 절편(편향) 5000
price = (area * 450 + 5000 + rng.normal(0, 2500, n)).round(-1)

# 특이 매물 5건 생성
special = rng.choice(n, 5, replace=False) # replace = False 같은 번호를 중복해서 뽑지 않는다.
price[special] += rng.uniform(15000, 25000, 5).round(-1)

office = pd.DataFrame({'area':area, "price": price})
office.describe().round(1)
````

**주의:** 셀 5를 두 번 실행해도 결과는 같아요. 칸 안에서 `default_rng(40)`으로 처음부터 다시 만들기 때문이에요. 대신 `rng` 줄을 빼고 아래 줄만 다시 실행하면 난수가 이어서 나와 값이 달라져요.

## 5. 선형회귀: 직선 y = wx + b 찾기

### 수업 셀 8: 학습

**나와야 하는 것:**

```text
기울기 w = 439.46 정답: 450
절편 b = 5926.68 정답: 5000
```

````python
# [셀 8] 선형회귀 학습
from sklearn.linear_model import LinearRegression

X = office[["area"]]   # 입력: 표(2차원), 대괄호 두 겹
y = office["price"]    # 정답: 열 하나, 대괄호 한 겹

# 선형 회귀 모델 만들기
model = LinearRegression()
# 학습 : X(입력값), y(정답)
model.fit(X, y)

print(f"기울기 w = {model.coef_[0].round(2)} 정답: 450")
print(f"절편 b = {model.intercept_.round(2)} 정답: 5000")
````

### 수업 셀 10: 모델의 직선 그리기

**나와야 하는 것:** 점 200개를 가로지르는 빨간 직선 하나와 범례 `실제 시세`, `모델의 직선`. 수업 노트북에서는 `plt.legend()`가 빠져 범례가 안 나왔어요. 여기서는 넣었어요.

````python
# [셀 10] 모델의 직선 그리기
# 학습한 모델로 예측하기
y_pred = model.predict(X)

line = X.sort_values("area")   # 면적이 작은 것부터 줄 세워야 선이 한 줄로 그려진다
line_pred = model.predict(line)

plt.figure(figsize=(8,5))
plt.scatter(office['area'], office["price"], alpha=0.6, label="실제 시세") # 알파는 점의 투명도
plt.plot(line["area"], line_pred, color="red", label="모델의 직선")

plt.title("오피스텔 시세 200건")
plt.xlabel("면적")
plt.ylabel("가격")
plt.legend()
plt.show()
````

## 6. 채점: MAE, MSE, RMSE, R²

### 수업 셀 11

**나와야 하는 것:**

```text
MAE = 2,493
MSE = 16,240,971
RMSE = 4,030
R_SCORE=0.803
```

````python
# [셀 11] 평가 지표
from sklearn.metrics import mean_absolute_error, mean_squared_error, root_mean_squared_error, r2_score
# MAE: 오차 절댓값의 평균 => 평균적으로 얼마만큼 틀리나?
print(f"MAE = {mean_absolute_error(y,y_pred):,.0f}")
# MSE: 오차 제곱의 평균 (단위 제곱)
print(f"MSE = {mean_squared_error(y,y_pred):,.0f}")
# RMSE: 오차 제곱 평균의 루트 (원래 단위)
print(f"RMSE = {root_mean_squared_error(y,y_pred):,.0f}")
# R SCORE: 1에 가까울수록 좋다 (평균으로 찍은 것보다 못하면 음수)
print(f"R_SCORE={r2_score(y, y_pred):,.3f}")
````

### 수업 셀 13: 고친 코드

계산 세 줄을 `for`문 안으로 들여 쓰고 `y_pred`를 `y_hat`으로 바꿨어요.

**나와야 하는 것:**

```text
전체 200건 MAE 2,493 | RMSE 4,030
특이매물 제외한 195 MAE 2,075 | RMSE 2,732
```

````python
# [셀 13] 전체 vs 특이매물 제외 비교
all_rows = np.ones(len(office), dtype=bool)
cases = [("전체 200건", all_rows), ("특이매물 제외한 195", normal)]

for name, mask in cases:
    y_true = y[mask]        # 실제가격
    y_hat = y_pred[mask]    # 예측가격

    mae = mean_absolute_error(y_true, y_hat)
    rmse = root_mean_squared_error(y_true, y_hat)
    print(f"{name} MAE {mae:,.0f} | RMSE {rmse:,.0f}")
````
